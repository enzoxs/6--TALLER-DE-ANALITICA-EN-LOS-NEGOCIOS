```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import matplotlib as plt
import scipy.stats as stats
import seaborn as sns

# fijar semilla para reproducibilidad
np.random.seed(42)

# numero de observaciones
n=100

# generar precios aleatorios entre 10 y 200
precios=np.linspace(10, 200, n)

# generar demanda a partir de la ecuacion de demanda con algo de ruido
a=1000 # demanda maxima teorica
b=5 # sensibilidad de la demanda al precio
error=np.random.normal(0, 50, n) # ruido con distribucion normal
demandas=a-b*precios+error # demanda real con ruido

# crear el dataframe con los datos generados
df=pd.DataFrame({'Precio': precios, 'Demanda': demandas})
df.head()
```
